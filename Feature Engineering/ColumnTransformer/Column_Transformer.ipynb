```python
# gender and city -> One Hot Encoding bz it is Nominal Column
# cough -> Ordinal Encoding because it is Ordinal Column
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import(
    OneHotEncoder,
    OrdinalEncoder
)
```

```python
from pandas.core.arrays import sparse
transformer=ColumnTransformer(transformers=[
    ('trf1',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('trf2',OneHotEncoder(sparse_output=False,drop='first',),['gender','city'])
])
```

```python
X_train
```

```python
New=transformer.fit_transform(X_train).astype(np.int8)
```

```python
New=pd.DataFrame(New,columns=transformer.get_feature_names_out())
```

```python
New['age']=df['age']
New['fever']=df['fever']
```
